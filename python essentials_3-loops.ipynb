## while loop

```python
i=0 ## simple while loop
while i<10:
    print("while condition is true",i)
    i += 1
```

```python
## while loop with break
i=0
while i<10:
    if i==4:
        break
    print(i)
    i +=1
    
```

```python
i=5 ## decrement
while i>=1:
    print(i)
    i=i-1
```

```python
i=1  # getting odd numbers
while i<=10:
    print(i)
    i=i+2
```

```python
i=0   # getting even numbers
while i<=10:
    print(i)
    i=i+2
```

```python
i=1   # getting even numbers
while i<=10:
    if i % 2 ==0:
        print(i)
    i=i+1
```

```python
# while loop with continue statement
i=0
while i<6:
    i+=1
    if i==3:
        continue
    print(i)
```

```python
# while loop with else statement.
i=7
while i<6:
    print(i)
    i+=1
else:
    print('while loop condition is false')
```

## filter

```python
seq =[0,1,2,3,5,8,13] # list contains both odd and even nos.

result= filter(lambda x:x%2!=0, seq) # result contains odd nos.
print(list(result))

result= filter(lambda x:x%2==0, seq) # result contains even nos.
print(list(result))
```

## map

```python
# add two lists using map and lambda

number1 = [1,2,3]
number2 = [4,5,6]

result= map(lambda x,y: x+y, number1, number2)
print(list(result))
```
