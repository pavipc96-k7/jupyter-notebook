#### get input for number a and b. print the number between a and b

### count the number which are divisible by 3 and 5. range 1-100

```python
count=0
for i in range(1,101):
    if(i%3==0 and i%5==0):
        count=count+1
print(count)
```

### write a program to display the cube of the number up to an integer.

```python
n=int(input("enter the number:"))
for i in range(1,n+1):
    cube=i*i*i
    print("the number is:",i and "the cube of the number is:",cube)
```
