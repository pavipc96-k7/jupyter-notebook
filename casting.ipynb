## get integer input for variable  a, b, c

multiply all 3 variables(a*b*c)
add all 3 variables(a+b+c)
divide the multiplied value by added values

```python
a=int(input())
b=int(input())
c=int(input())
d=a*b*c
print(d)
e=a+b+c
print(e)
f=d/e
print(f)
```
